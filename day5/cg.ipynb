```python
from fastapi import FastAPI
from pydantic import BaseModel
import nest_asyncio
import uvicorn

# Allow server to run in Colab
nest_asyncio.apply()

# Create FastAPI app
app = FastAPI()


# Pydantic model
class Student(BaseModel):
    name: str
    age: int
    branch: str


# Simple database
students = {}


# GET - Get all students
@app.get("/students")
def get_students():
    return students


# GET - Get student by ID
@app.get("/students/{student_id}")
def get_student(student_id: int):
    return students.get(student_id, "Student not found")


# GET - Query parameter
@app.get("/search")
def search_student(branch: str):
    return {
        "branch": branch,
        "message": "Searching students"
    }


# POST - Add student
@app.post("/students/{student_id}")
def add_student(student_id: int, student: Student):
    students[student_id] = student.model_dump()

    return {
        "message": "Student added",
        "student": students[student_id]
    }


# PUT - Update student
@app.put("/students/{student_id}")
def update_student(student_id: int, student: Student):
    students[student_id] = student.model_dump()

    return {
        "message": "Student updated",
        "student": students[student_id]
    }


# DELETE - Delete student
@app.delete("/students/{student_id}")
def delete_student(student_id: int):
    student = students.pop(student_id, None)

    return {
        "message": "Student deleted",
        "student": student
          }


# Run server

import threading

def run_server():
    uvicorn.run(
        app,
        host="0.0.0.0",
        port=8000
    )

server_thread = threading.Thread(
    target=run_server,
    daemon=True
)

server_thread.start()

print("FastAPI server started!")
print("Running on port 8000")
```
